```python
import pandas as pd
import matplotlib.pyplot as plt

# Ruta a tu CSV
csv_path = "/content/drive/MyDrive/runs_uav/monet_pop_p2_1024/results.csv"

# Cargar CSV
df = pd.read_csv(csv_path)

# Graficar
plt.figure(figsize=(10,6))
plt.plot(df["epoch"], df["metrics/precision(B)"], label="Precision", marker="o")
plt.plot(df["epoch"], df["metrics/recall(B)"], label="Recall", marker="o")
plt.plot(df["epoch"], df["metrics/mAP50-95(B)"], label="mAP50-95", marker="o")

plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title("Evolución de Precision, Recall y mAP50-95")
plt.legend()
plt.grid(True)
plt.show()

# Mostrar la mejor epoch por mAP50-95
best_idx = df["metrics/mAP50-95(B)"].idxmax()
print("⚡ Mejor epoch:", df.loc[best_idx, "epoch"])
print("mAP50-95:", df.loc[best_idx, "metrics/mAP50-95(B)"])
print("Precision:", df.loc[best_idx, "metrics/precision(B)"])
print("Recall:", df.loc[best_idx, "metrics/recall(B)"])
```
